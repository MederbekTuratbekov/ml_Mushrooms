```python
# 1. Заменили '?' на NaN
df['stalk-root'] = df['stalk-root'].replace('?', np.nan)
```

```python
for i in df.columns:
    print(f"{i}: {(df[i] == '?').sum()}")
```

Обучение

```python
df['class'] = df['class'].map({'p': 1, 'e': 0})
```

```python
df.head(3)
```

```python
x = df.drop(columns=['class'])
y = df['class']
```

```python
# 2. Сплит до кодирования (нет утечки данных)
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
# 3. Заполнение NaN модой из x_train
mode_val = x_train['stalk-root'].mode()[0]
x_train['stalk-root'] = x_train['stalk-root'].fillna(mode_val)
x_test['stalk-root'] = x_test['stalk-root'].fillna(mode_val)
```

```python
# 4. Кодирование категориальных признаков
ohe = OneHotEncoder(drop='first', sparse_output=False, handle_unknown='ignore')
x_train_encoded = ohe.fit_transform(x_train)
x_test_encoded = ohe.transform(x_test)
```

```python
x_train_encoded
```

```python
# 5. Обучение моделей НА encoded ДАННЫХ
model_lg = LogisticRegression(max_iter=1000)
model_tree = DecisionTreeClassifier(max_depth=5, random_state=42)
model_rf = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=42)
```

```python
model_lg.fit(x_train_encoded, y_train)
model_tree.fit(x_train_encoded, y_train)
model_rf.fit(x_train_encoded, y_train)
```

```python
y_predict_lg = model_lg.predict(x_test_encoded)
y_predict_tree = model_tree.predict(x_test_encoded)
y_predict_rf = model_rf.predict(x_test_encoded)
```

```python
print('accuracy_score:')
print(f'Model Log: {accuracy_score(y_test, y_predict_lg)}')
print(f'Model Tree: {accuracy_score(y_test, y_predict_tree)}')
print(f'Model Rf: {accuracy_score(y_test, y_predict_rf)}')
```

```python
print('precision_score:')
print(f'Model Log: {precision_score(y_test, y_predict_lg)}')
print(f'Model Tree: {precision_score(y_test, y_predict_tree)}')
print(f'Model Rf: {precision_score(y_test, y_predict_rf)}')
```

```python
print('recall_score:')
print(f'Model Log: {recall_score(y_test, y_predict_lg)}')
print(f'Model Tree: {recall_score(y_test, y_predict_tree)}')
print(f'Model Rf: {recall_score(y_test, y_predict_rf)}')
```

```python
print('f1_score:')
print(f'Model Log: {f1_score(y_test, y_predict_lg)}')
print(f'Model Tree: {f1_score(y_test, y_predict_tree)}')
print(f'Model Rf: {f1_score(y_test, y_predict_rf)}')
```

```python
print('classification_report:')
print(f'Model Log: {classification_report(y_test, y_predict_lg)}')
print(f'Model Tree: {classification_report(y_test, y_predict_tree)}')
print(f'Model Rf: {classification_report(y_test, y_predict_rf)}')
```
